# Social Media Data Analysis Walkthrough
This notebook provides an interactive, step-by-step walkthrough of our analysis pipeline.

```python
import sys
import os
# Add the parent directory to the path so we can import our modules
sys.path.append(os.path.abspath('..'))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import generate_synthetic_data
from src.data_cleaning import clean_dataset
from src.sentiment_analysis import analyze_sentiment
from src.engagement_analysis import calculate_engagement_rate, analyze_correlations

# Setup seaborn for inline plots
sns.set_theme(style="whitegrid")
%matplotlib inline
```

## 1. Data Loading and Generation

```python
df = generate_synthetic_data(num_rows=500)
df.head()
```

## 2. Data Cleaning
We remove URLs, mentions, emojis, and handle any missing values.

```python
df_clean = clean_dataset(df)
df_clean[['text', 'cleaned_text']].head()
```

## 3. Sentiment Analysis
We apply VADER sentiment analysis to classify each post.

```python
df_analyzed = analyze_sentiment(df_clean)

# Show sentiment distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=df_analyzed, x='sentiment_label')
plt.title('Sentiment Distribution')
plt.show()
```

## 4. Engagement Analysis
We calculate engagement rates and analyze correlations.

```python
df_engaged = calculate_engagement_rate(df_analyzed)

correlations = analyze_correlations(df_engaged)
plt.figure(figsize=(8, 6))
sns.heatmap(correlations, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```
